# Content-Based Filtering Recommendation System (From Scratch)

In this notebook, we build a content-based recommendation system
that recommends items to users based on item features and user preferences.


```python
import numpy as np
```

```python
# Rows = items, Columns = features
# Example: [Action, Comedy, Drama]
item_features = np.array([
    [1, 0, 0],  # Movie 0
    [1, 1, 0],  # Movie 1
    [0, 1, 0],  # Movie 2
    [0, 0, 1],  # Movie 3
    [1, 0, 1]   # Movie 4
])
```

```python
# User preference for genres
user_profile = np.array([0.8, 0.3, 0.1])
```

## Content-Based Filtering Intuition

Recommend items similar to what the user likes.

Score = similarity(user_profile, item_features)

```python
# Compute Similarity Scores
def compute_scores(user_profile, item_features):
    return item_features @ user_profile
```

```python
# Generate Recommendations
scores = compute_scores(user_profile, item_features)
scores
```

```python
# Top Recommended Items
recommended_items = np.argsort(scores)[::-1]
recommended_items
```

## Handling Already Consumed Items

In practice, already watched items are filtered out
before recommending new ones.


```python
# Finding Related Items
def cosine_similarity(A, B):
    return np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))
```

```python
item_id = 0
similarities = []

for i in range(len(item_features)):
    sim = cosine_similarity(item_features[item_id], item_features[i])
    similarities.append(sim)

similarities
```

## Content-Based vs Collaborative Filtering

Content-Based:
- Uses item features
- No cold start for users
- Limited discovery

Collaborative Filtering:
- Uses user behavior
- Better discovery
- Cold start problem
