# Reinforcement Learning with Neural Networks

In this notebook, we use a neural network to approximate
the state-value function instead of using a lookup table.
This approach enables reinforcement learning in large or continuous state spaces.


## Why Use Neural Networks in RL?

- Large state spaces cannot be stored in tables
- Neural networks generalize across states
- Used in Deep Q-Learning, AlphaGo, robotics


```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers
```

```python
# Define Environment
states = np.array([[0], [1], [2]])  # Simple 1D state
gamma = 0.9
```

```python
# Reward Function
def reward(state):
    return 1 if state == 2 else 0
```

```python
# build value network
model = tf.keras.Sequential([
    layers.Dense(16, activation="relu", input_shape=(1,)),
    layers.Dense(16, activation="relu"),
    layers.Dense(1)  # V(s)
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
    loss="mse"
)
```

```python
# Bellman Targets
def bellman_target(state):
    next_state = min(state + 1, 2)
    return reward(state) + gamma * model.predict(np.array([[next_state]]), verbose=0)[0][0]
```

```python
#  training loop
for episode in range(200):
    X_train = []
    y_train = []

    for s in [0, 1]:
        X_train.append([s])
        y_train.append(bellman_target(s))

    model.fit(np.array(X_train), np.array(y_train),
              epochs=1, verbose=0)
```

```python
#  learned value function
for s in [0, 1, 2]:
    v = model.predict(np.array([[s]]), verbose=0)[0][0]
    print(f"V({s}) ≈ {v:.3f}")
```

## Relation to Deep Reinforcement Learning

- This is value function approximation
- Extends to Q-learning → Deep Q-Networks (DQN)
- Basis of modern RL algorithms


## Key Takeaways

- Neural networks replace value tables
- Bellman equation provides training targets
- Enables RL in complex environments
