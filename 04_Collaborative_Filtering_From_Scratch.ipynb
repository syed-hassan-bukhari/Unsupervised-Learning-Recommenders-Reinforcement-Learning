# Collaborative Filtering Recommendation System (From Scratch)

In this notebook, we build a collaborative filtering recommendation system
from scratch using matrix factorization.

The goal is to learn user and item features that can predict user ratings
for unseen items.


```python
import numpy as np
```

```python
# Rows = users, Columns = items
# 0 means no rating
R = np.array([
    [5, 4, 0, 0],
    [3, 0, 0, 2],
    [4, 0, 0, 5],
    [0, 3, 4, 0],
    [0, 0, 5, 4]
])
```

## Collaborative Filtering Intuition

Users with similar preferences tend to like similar items.

We factorize the rating matrix R into:
- User feature matrix X
- Item feature matrix W

Prediction = X · Wᵀ


```python
num_users, num_items = R.shape
num_features = 2

X = np.random.rand(num_users, num_features)
W = np.random.rand(num_items, num_features)
```

```python
# Prediction Function
def predict(X, W):
    return X @ W.T
```

```python
# Cost Function with Regularization
def compute_cost(R, X, W, lambda_):
    cost = 0
    for i in range(num_users):
        for j in range(num_items):
            if R[i, j] != 0:
                error = X[i] @ W[j] - R[i, j]
                cost += error ** 2
    reg = lambda_ * (np.sum(X**2) + np.sum(W**2))
    return cost + reg
```

```python
# Gradient Descent
def gradient_descent(R, X, W, alpha, lambda_, iterations):
    for _ in range(iterations):
        for i in range(num_users):
            for j in range(num_items):
                if R[i, j] != 0:
                    error = X[i] @ W[j] - R[i, j]
                    X[i] -= alpha * (error * W[j] + lambda_ * X[i])
                    W[j] -= alpha * (error * X[i] + lambda_ * W[j])
    return X, W
```

# Train the model

```python
alpha = 0.01
lambda_ = 0.1
iterations = 500

X, W = gradient_descent(R, X, W, alpha, lambda_, iterations)
```

```python
predicted_ratings = predict(X, W)
predicted_ratings
```

```python
# EXAMPLE
user_id = 0
recommended_item = np.argmax(predicted_ratings[user_id])
recommended_item
```
