# Reinforcement Learning Fundamentals

This notebook introduces the core components of Reinforcement Learning:
states, actions, rewards, policies, value functions, and the Bellman equation.


## Core Components of Reinforcement Learning

- State (s)
- Action (a)
- Reward (r)
- Policy (π)
- Value Function (V)
- Discount Factor (γ)


```python
import numpy as np
```

```python
# Define Environment
states = [0, 1, 2]   # 0 = start, 2 = goal
actions = ["left", "right"]

gamma = 0.9
```

```python
# Define Reward Function
def reward(state, action):
    if state == 1 and action == "right":
        return 1
    return 0
```

```python
# Define Policy
policy = {
    0: "right",
    1: "right",
    2: "right" # Added policy for state 2
}
```

```python
# State Transition Function
def transition(state, action):
    if action == "right":
        return min(state + 1, 2)
    return max(state - 1, 0)
```

**Return**

The return Gₜ is the total discounted reward:

Gₜ = rₜ₊₁ + γrₜ₊₂ + γ²rₜ₊₃ + ...


```python
# State Value Function
def compute_state_value(state, policy, gamma):
    value = 0
    discount = 1
    for _ in range(10):
        action = policy[state]
        r = reward(state, action)
        state = transition(state, action)
        value += discount * r
        discount *= gamma
    return value
```

```python
# Compute Values for All States
V = {s: compute_state_value(s, policy, gamma) for s in states}
V
```

## Bellman Equation

V(s) = R(s,a) + γ · V(s')

The value of a state equals immediate reward plus
the discounted value of the next state.


```python
# Bellman Update
def bellman_update(state, gamma):
    action = policy[state]
    return reward(state, action) + gamma * V.get(transition(state, action), 0)
```

```python
for s in states:
    V[s] = bellman_update(s, gamma)

V
```
